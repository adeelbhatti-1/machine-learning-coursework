# Gini index calculation

```python
from __future__ import division
import numpy as np
```

```python
elements = ['A', 'A', 'A', 'C', 'B', 'C']
def counts(elements):
    classes = {}
    for element in elements:
        if element in classes:
            classes[element] += 1
        else:
            classes[element] = 1
    return [classes[e] for e in classes]
counts(elements)
```

```python
def gini(elements):
    cts = counts(elements)
    n = sum(cts)
    return 1 - sum([p_i**2/n**2 for p_i in cts])
gini(elements)
```

```python
def entropy(elements):
    if len(elements) == 0:
        return 0
    cts = counts(elements)
    n = sum(cts)
    # Compute probabilities (p_i)
    probs = [count / n for count in cts]
    # Apply standard entropy formula: -sum(p_i * log2(p_i))
    return -sum(p * np.log2(p) for p in probs)
```

```python
for i in range(len(elements)):
    print("************")
    left = elements[:i]
    right = elements[i:]
    print(left, right)
    weighted_gini = 1/len(elements)*(gini(left)*len(left) + gini(right)*len(right))
    print("Weighted Gini:", weighted_gini)
    weighted_entropy = 1/len(elements)*(entropy(left)*len(left) + entropy(right)*len(right))
    print("Weighted entropy:", weighted_entropy)
```

```python
# Calculate Gini Index and Entropy using the standard formulas
gini_index = gini(elements)
entropy_val = entropy(elements)

print(f"Dataset: {elements}")
print(f"Gini Index: {gini_index:.6f}")
print(f"Entropy: {entropy_val:.6f}")
```
